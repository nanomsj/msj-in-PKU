## Ito isometry

### 1. (20pts)
Find the variance of the following random variables by applying the Ito isometry 

$$
\Eof{\left(\int_a^b \varphi_s dB_s\right)^2} = \int_a^b \Eof{\varphi_s^2}ds
$$

- (a) $X = \dint_0^t B_s^2 dB_s$. In this case, $\varphi_s = B_s^2$.
- (b) $X = \int_a^b \sqrt t B_t dB_t$. In this case, $\varphi_t = \sqrt t B_t$.

### <font color=blue> Solution 1(a). </font>

Since the expectation of X is 0, and for $Y \sim N(0,t), E Y^4= 3t^2$ (can be proved by Gaussian Integral easily), thus

$$
\Var X =\Eof{\left(\int_0^t B_s^2 dB_s\right)^2} = \int_0^t \Eof{(B_s^2)^2}ds =  \int_0^t \Eof{B_s^4}ds = \int_0^t 3s^2ds=t^3
$$


### <font color=blue> Solution 1(b). </font>

The expectation of X is 0. 

$E B_t^2 = t$, so $\Eof{t B_t^2} = t \Eof{B_t^2} = t^2$

$$
\Var X = \Eof{\left(\int_a^b \sqrt t B_t dB_t\right)^2} = \int_a^b \Eof{(\sqrt t B_t)^2}dt = \int_a^b \Eof{t B_t^2}dt = \int_a^b t^2dt=\frac{b^3-a^3}{3}
$$




## Apply Ito's formula to evaluate stochastic integral

### 2. (20pts)
Apply Ito's formula to evaluate the following stochastic integrals.

- (a) $\dint_0^t B_s e^{B_s} dB_s$ 
- (b) $\dint_0^t \frac 1{1 + B_s^2} dB_s$

### <font color=blue> Solution 2(a). </font>

Ito's Formula: $df(B_t)=f'(B_t)\,dB_t+\frac12 f''(B_t)\,dt.$

Since $\int x e^x dx= xe^x-e^x + C$, 

$$
\dint_0^t B_s e^{B_s} dB_s = \int_0^t d(B_s e^{B_s}-e^{B_s})- \frac12\int_0^t (B_se^{B_s}+ e^{B_s})  ds= B_t e^{B_t}-e^{B_t}+1-\frac12\int_0^t (B_se^{B_s}+ e^{B_s}) ds
$$

### <font color=blue> Solution 2(b). </font>

$$
\dint_0^t \frac 1{1 + B_s^2} dB_s = \int_0^t d \arctan(B_s)-\int_0^t \frac12 \frac{-2B_s}{(1+B_s^2)^2} ds = \arctan(B_t) +\int_0^t \frac{B_s}{(1+B_s^2)^2} ds .
$$

## Apply Ito's formula to solve SDE

### 3. (10pts)

Let $X_t$ satisfy the SDE for Ornstein-Uhlenbeck process

$$
dX_t = \lambda(m - X_t) dt + \sigma dB_t, \quad X_0 = x > 0,
$$

where $\lambda > 0$, $m$ and $\sigma$ are constants. 

Apply Ito's formula to $e^{\lambda t}X_t$ and show that the solution is given by

$$
X_t = e^{-\lambda t} x + \left(1 - e^{-\lambda t} \right) m + \sigma \int_0^t e^{-\lambda(t - s)} dW_s.
$$


### <font color=blue> Solution 3. </font>

According to Ito's formula, 
$$
\begin{aligned}
d(e^{\lambda t}X_t)=\left(e^{\lambda t}\right)dX_t+\left(\lambda e^{\lambda t}X_t\right)dt
=& \left(e^{\lambda t}\right)(\lambda(m - X_t) dt + \sigma dB_t)+\left(\lambda e^{\lambda t}X_t\right)dt\\
=& \left(e^{\lambda t}\right)(\lambda m dt + \sigma dB_t)\\
\end{aligned}
$$

Integrating both side from 0 to t,
$$
e^{\lambda t}X_t- x = \int_0^t \left(e^{\lambda s}\right)(\lambda m ds + \sigma dB_s)
=(e^{\lambda t}-1)m+\sigma \int_0^t e^{\lambda s}dB_s
$$

Multiply both side by $e^{-\lambda t}$, we get the solution: 

$$
X_t = e^{-\lambda t} x + \left(1 - e^{-\lambda t} \right) m + \sigma \int_0^t e^{-\lambda(t - s)} dB_s.
$$

Substitute B=W (both are standard brownian motion), we get the same equation as above.


## Simulating the Black-Scholes model

### 4. (20pts)

With 100 steps within the time interval $[0,1]$, simulate 20,000 sample paths for the Black-Scholes model 

$$
dX_t = r X_t dt + \sigma X_t dB_t
$$

by the Euler-Maruyama scheme with $X_0 = 100$. The parameters are given by $r = 2\%$ and $\sigma = 25\%$.

- (a) Plot the histograms of $X_t$ for $t = 0.4, 1$.
- (b) Evaluate the expectation $e^{-rT}\Eof{g(X_T)}$ by taking sample mean from using the simulated samples in (a), where $T = 1$ and $g$ is given by 
$$
g(x) = \left\{\begin{array}{ll}
5 & \text{ if } x > 101, \\
-4 & \text{ if } x \leq 101.
\end{array}\right.
$$
Plot the histogram of $g(X_1)$. 

Note: In `python`, the function $g$ can be coded as `5*(x>101) - 4*(x<=101)`. 

```python
# as always, import required modules and functions
import numpy as np
from numpy import sqrt, log, exp
import matplotlib.pyplot as plt
import scipy.stats as ss
from scipy.stats import norm
import seaborn as sns
```

```python
#(4a)
# set random seed
np.random.seed(1609)

# Parameters of the Black-Scholes model
r, sigma = 0.02,  0.25
x0, T = 100, 1

# number of paths and number of time steps
n_sim, n_steps = 20_000, 100

# initialize X
X = np.ones([n_sim, n_steps+1])*x0

# Euler-Maruyama scheme
dt = T/n_steps
for i in range(n_steps):
    db = norm.rvs(size=n_sim)
    db = db - db.mean()
    db = db/db.std()
    X[:,i+1] = X[:,i] + r*X[:,i]*dt + sigma*X[:,i]*np.sqrt(dt)*db # change here qwq

# plot
plt.figure(figsize=(12, 8))
t = np.arange(0, T+dt, dt)

# histograms at different times
plt.subplot(2, 2, 1)
time = 0.4
sns.histplot(X[:,t==time], bins=30, stat='density', label=f'$t={time}$', element='step')
sns.histplot(X[:,-1], bins=30, stat='density', label=f'$t={T}$', color='lightgreen', element='step')
plt.legend()
```

```python
# 4b
g_X1 = 5 * (X[:, -1] > 101) - 4 * (X[:, -1] <= 101)
a= exp(-r * T) * np.mean(g_X1)

print("Discounted expectation:", a) 
```

```python
# histogram of g(X1)
plt.hist(g_X1,bins=[-4.5, -3.5, 4.5, 5.5],density=True) 

plt.xticks([-4, 5])
plt.xlabel("$g(X_1)$")
plt.ylabel("Density")
plt.title("Histogram of $g(X_1)$")
plt.show() 
```
